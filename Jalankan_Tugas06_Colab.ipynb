> Jangan gunakan repo privat; Colab tidak bisa meng-clone repo privat tanpa token.

```python
REPO_URL = "https://github.com/USERNAME/REPO.git"  # GANTI DENGAN URL REPO ANDA
print(REPO_URL)
```

```python
import os, pathlib, subprocess

repo_name = REPO_URL.rstrip('/').split('/')[-1].replace('.git', '')
os.chdir('/content')
if pathlib.Path(repo_name).exists():
    subprocess.run(['git', '-C', repo_name, 'pull'], check=True)
else:
    subprocess.run(['git', 'clone', REPO_URL], check=True)
os.chdir(repo_name)
print('Repo aktif:', os.getcwd())
print(subprocess.run(['git', 'log', '--oneline', '-3'], capture_output=True, text=True).stdout)
```
